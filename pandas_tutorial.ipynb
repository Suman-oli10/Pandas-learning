**Filtering Data**

```python
bios=pd.read_csv("./data/bios.csv")
```

```python
#with loc
bios.loc[bios['height_cm']>220,['name','height_cm']]
```

```python
#Without loc
bios[bios['height_cm'] > 220][['name','height_cm']]
```

In such way we can filter any data from huge dataframe

```python
#Filtering the data using multiple conditions
bios[(bios['height_cm']>215) & (bios['born_country']=='USA')][['name','height_cm','born_country']]
```

```python
bios[bios['name'].str.contains('suman',case=False)]
```

```python
bios[bios['born_country'] == 'NEP'][['name','born_country','born_city']]
```
