```python
bios_new=bios.copy()
```

```python
bios_new['first_name']=bios_new['name'].str.split(' ').str[0]
bios_new.head()
```

```python
bios_new.query('first_name=="Keith"') #WE can see keith only if we scroll right in the first_name section
```

```python
bios_new['DOB']=pd.to_datetime(bios_new['born_date'])#Conveting the born_date to 'to_datetime' so that we can point out born_year
bios_new['born_year']=bios_new['DOB'].dt.year
bios_new[['name','born_year']].head()
```

```python
bios_new.to_csv('./data/bios_new',index=False) #Making a new csv dataframe: bios_new stored in data folder
```

```python
#Adding a new column(height_category) in bios dataframe
bios_new['height_category']=bios_new['height_cm'].apply(lambda x: 'Short' if x < 170 else 'Average' if x < 185 else 'Tall')
bios_new.head(8)

#Scroll to the right to see the result and learn the pattern how we calculate height_categorya and born_year.
```

```python
#In this way we can apply conditions too using function:

def athelete_category(row):
    if row['height_category']=='short' or row['weight_kg'] < 70:
        return 'Lightweight'
    elif row['height_category']=='Average' or row['weight_kg'] < 75:
         return 'Middleweight'
    else:
        return 'Heavyweight'

bios_new['category']=bios_new.apply(athelete_category, axis=1)
bios_new
```

**Merging and Concatenating Data**

```python
#Reading a noc_regions file from data folder
noc=pd.read_csv('./data/noc_regions.csv')
noc.head()
```

```python
#Merging bios_new file with noc file and store them in bios_new_new file
bios_new_new=bios_new.copy()
bios_new_new=pd.merge(bios_new, noc,left_on='born_country',right_on='NOC' ,how='left')
bios_new_new.head()

#we merge (born_country) of bios_new dataframe with 'NOC' of noc dataframe. 
#we use (left_on = born_country) and (right_on = NOC) in this code because these two dataframehave
# same values in these both columns that helps us to merge it taking the values as reference.
```

```python
bios_new_new.rename(columns={'region':'born_country_full'},inplace=True)
bios_new_new.head()
```

```python
#Creating a new df so that we can concat
usa=bios[bios['born_country']=='USA'].copy()
gbr=bios[bios['born_country']=='GBR'].copy()
```

```python
#Concatenating usa and gbr dataframe
new_df=pd.concat([usa,gbr])
```
